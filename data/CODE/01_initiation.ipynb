```python
# Mapped each image to its disease label
import os
import pandas as pd

# Load metadata CSV
df = pd.read_csv(r"K:\I.T\Skin Disease Detection\data\HAM10000_metadata.csv")

# Define image folder path
image_folder = r"K:\I.T\Skin Disease Detection\data\HAM10000_images"

# Add a 'path' column by combining image_id with the folder path
df["path"] = df["image_id"].map(lambda x: os.path.join(image_folder, x + ".jpg"))

# Quick check
print(df.head())
```
